```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
conn = sqlite3.connect('/content/SQLite.db')
```

```python
query = "SELECT * FROM dashboard_data;"
df = pd.read_sql_query(query, conn)
conn.close()
```

```python
print("### Portfolio Summary Statistics\n")
print(f"- **Total Loans Analyzed:** {len(df):,}")
print(f"- **Average Credit Score:** {df['credit_score'].mean():.1f}")
print(f"- **Median Annual Income:** ${df['annual_income'].median():,.2f}")
print(f"- **Average Debt-to-Income (DTI):** {df['debt_to_income_percentage'].mean():.2f}%\n")
```

```python
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x='debt_to_income_percentage',
    y='credit_score',
    hue='home_ownership',
    alpha=0.6
)
plt.title('Credit Score vs. Debt-to-Income Percentage')
plt.xlabel('Debt-to-Income Ratio (%)')
plt.ylabel('Credit Score')
plt.savefig('dti_vs_credit_score.png', dpi=300, bbox_inches='tight')
plt.close()
```

```python
plt.figure(figsize=(12, 6))
purpose_risk = df.groupby('purpose').agg(
    avg_loan=('current_loan_amo', 'mean'),
    avg_problems=('number_of_credit', 'mean')
).reset_index()

sns.barplot(
    data=purpose_risk.sort_values(by='avg_loan', ascending=False),
    x='avg_loan',
    y='purpose',
    palette='viridis'
)
plt.title('Average Loan Amount requested by Purpose')
plt.xlabel('Average Loan Amount ($)')
plt.ylabel('Loan Purpose')
plt.savefig('loan_amount_by_purpose.png', dpi=300, bbox_inches='tight')
plt.close()

print("Charts successfully saved as 'dti_vs_credit_score.png' and 'loan_amount_by_purpose.png'!")
```
